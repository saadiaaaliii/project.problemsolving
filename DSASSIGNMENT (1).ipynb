```python
gcf_long = gcf_clean.melt(
    id_vars=["Country Name", "Country Code"],
    value_vars=gcf_clean.columns[4:],
    var_name="Year",
    value_name="GCF_Growth"
)

gcf_long["Year"] = (
    gcf_long["Year"].astype(str).str[:4].astype(int)
)

gcf_long["GCF_Growth"] = pd.to_numeric(
    gcf_long["GCF_Growth"],
    errors="coerce"
)

print("Available GCF values:",
      gcf_long["GCF_Growth"].notna().sum())
```

```python
df_final = df_numeric.merge(
    gcf_long[["Country Code", "Year", "GCF_Growth"]],
    on=["Country Code", "Year"],
    how="left",
    validate="one_to_one"
)

df_final[["Exports", "GDP_Growth", "GCF_Growth"]].isna().sum()
```

```python
df_final[
    ["Exports", "GDP_Growth", "GCF_Growth"]
].dropna().shape
```

```python
analysis = df_final.dropna(
    subset=["Exports", "GDP_Growth", "GCF_Growth"]
).copy()

print("Observations:", len(analysis))

print(
    analysis[["Exports", "GDP_Growth", "GCF_Growth"]]
    .corr()
)
```

```python
median_gcf = analysis["GCF_Growth"].median()

analysis["GCF_Group"] = analysis["GCF_Growth"].apply(
    lambda x: "High GCF Growth" if x >= median_gcf
    else "Low GCF Growth"
)

print("Median GCF growth:", median_gcf)

print(analysis["GCF_Group"].value_counts())
```

```python
for group, data in analysis.groupby("GCF_Group"):
    correlation = data["Exports"].corr(data["GDP_Growth"])

    print(group)
    print("Observations:", len(data))
    print("Correlation:", round(correlation, 4))
    print()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.scatter(
    complete["Exports"],
    complete["GDP_Growth"],
    alpha=0.4
)

plt.title("Exports and GDP Growth Show a Weak Positive Relationship")
plt.xlabel("Exports of Goods and Services (% of GDP)")
plt.ylabel("GDP Growth (Annual %)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
selected_countries = [
    "Pakistan",
    "India",
    "Bangladesh",
    "China",
    "Norway"
]

comparison = complete[
    complete["Country Name"].isin(selected_countries)
].copy()

comparison.groupby("Country Name")["Exports"].mean()
```

```python
country_avg = comparison.groupby("Country Name")["Exports"].mean()

plt.figure(figsize=(10, 5))

country_avg.sort_values().plot(kind="bar")

plt.title("Average Exports as a Share of GDP: Selected Countries")
plt.xlabel("Country")
plt.ylabel("Exports (% of GDP)")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

higher exports share doesnt mean higher gdp growth.

```python
import matplotlib.pyplot as plt

groups = ["Low GCF Growth", "High GCF Growth"]
correlations = [0.0980, 0.0534]

plt.figure(figsize=(8, 5))
plt.bar(groups, correlations)

plt.title("Exports–GDP Growth Correlation Is Weaker at Higher GCF Growth")
plt.xlabel("GCF Growth Group")
plt.ylabel("Correlation")
plt.ylim(0, 0.12)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

exports–GDP growth varies somewhat across investment groups, but remains weak in both groups. This does not indicate a causal effect,its just an association.
